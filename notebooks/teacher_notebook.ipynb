## Exercise 2 Solution: Image Validation

### Teaching Points:
- Dictionary comprehensions
- Boolean logic
- Function reusability

### Discussion Points:
- Why use a dictionary instead of a list?
- How could we extend this to check file size?
- What about checking if files actually exist?

```python
# SOLUTION TO EXERCISE 2
# ======================

def validate_image_files(filenames):
    """
    Check which filenames are valid image files.
    
    Args:
        filenames (list): List of filename strings
        
    Returns:
        dict: Dictionary mapping filename to True/False
    """
    # Solution using dictionary comprehension
    return {filename: is_allowed_file(filename) for filename in filenames}

# Test the function
test_files = ["photo.jpg", "doc.pdf", "image.png", "video.mp4", "picture.gif"]
result = validate_image_files(test_files)

print("Validation results:")
for filename, is_valid in result.items():
    status = "✓" if is_valid else "✗"
    print(f"  {status} {filename}")

# Count valid files
valid_count = sum(result.values())
print(f"\nValid image files: {valid_count}/{len(test_files)}")
```

### Alternative Solutions:

```python
# Alternative 1: Using a regular loop (more beginner-friendly)
def validate_image_files_v2(filenames):
    """Loop-based version (easier to understand)."""
    results = {}
    for filename in filenames:
        results[filename] = is_allowed_file(filename)
    return results

# Alternative 2: Return only valid files
def get_valid_image_files(filenames):
    """Return list of only valid image filenames."""
    return [f for f in filenames if is_allowed_file(f)]

# Test alternatives
print("\nAlternative 1:", validate_image_files_v2(test_files))
print("\nAlternative 2:", get_valid_image_files(test_files))
```

## Exercise 3 Solution: Image Resizing Pipeline

### Teaching Points:
- Error handling with try/except
- Function composition
- Printing informative messages

### Extension Ideas:
- Add option to save resized image
- Support multiple resize strategies (crop, pad, etc.)
- Maintain aspect ratio information

```python
# SOLUTION TO EXERCISE 3
# ======================

def load_and_resize(image_path, max_size=256):
    """
    Load an image and resize it to maximum dimensions.
    
    Args:
        image_path (str): Path to the image file
        max_size (int): Maximum dimension (width or height)
        
    Returns:
        PIL.Image: The loaded and resized image
        
    Raises:
        Exception: If image cannot be loaded or processed
    """
    try:
        # Step 1: Load the image
        print(f"Loading image from: {image_path}")
        image = load_image_from_file(image_path)
        
        # Step 2: Print original size
        original_size = image.size
        print(f"Original size: {original_size[0]}x{original_size[1]} pixels")
        
        # Step 3: Resize the image
        resized_image = resize_image(image, max_size=max_size)
        
        # Step 4: Print new size
        new_size = resized_image.size
        print(f"New size: {new_size[0]}x{new_size[1]} pixels")
        
        # Calculate size reduction
        reduction = (1 - (new_size[0] * new_size[1]) / (original_size[0] * original_size[1])) * 100
        print(f"Size reduction: {reduction:.1f}%")
        
        # Step 5: Return the resized image
        return resized_image
        
    except FileNotFoundError:
        print(f"Error: File not found at {image_path}")
        raise
    except Exception as e:
        print(f"Error processing image: {e}")
        raise

# Test the function
print("="*60)
test_image = load_and_resize("images/sample_cat.jpg", max_size=200)
print("="*60)

# Display the result
plt.figure(figsize=(8, 6))
plt.imshow(test_image)
plt.axis('off')
plt.title(f"Resized Image ({test_image.size[0]}x{test_image.size[1]})")
plt.show()
```

### Bonus: Side-by-side comparison

```python
# Bonus: Compare original and resized
original = load_image_from_file("images/sample_cat.jpg")
resized = load_and_resize("images/sample_cat.jpg", max_size=200)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))
axes[0].imshow(original)
axes[0].set_title(f"Original ({original.size[0]}x{original.size[1]})")
axes[0].axis('off')

axes[1].imshow(resized)
axes[1].set_title(f"Resized ({resized.size[0]}x{resized.size[1]})")
axes[1].axis('off')

plt.tight_layout()
plt.show()
```

## Exercise 4 Solution: Caption Analysis

### Teaching Points:
- Text processing
- Using collections.Counter
- Set operations for filtering
- String methods (lower, split)

### Extension Ideas:
- Visualize word frequencies with a bar chart
- Find unique words per image
- Calculate caption similarity

```python
# SOLUTION TO EXERCISE 4
# ======================

def count_words(captions):
    """
    Count word frequency in a list of captions.
    
    Args:
        captions (list): List of caption strings
        
    Returns:
        Counter: Word frequency counter
    """
    # Define stop words to ignore
    stop_words = {'a', 'an', 'the', 'is', 'of', 'in', 'on', 'at', 'to', 'with'}
    
    # Step 1: Combine all captions
    all_text = ' '.join(captions)
    
    # Step 2: Split into words and convert to lowercase
    words = all_text.lower().split()
    
    # Step 3: Filter out stop words and non-alphabetic words
    filtered_words = [
        word for word in words 
        if word not in stop_words and word.isalpha()
    ]
    
    # Step 4: Count frequencies
    word_counts = Counter(filtered_words)
    
    return word_counts

# Test with sample images
print("Loading sample images and generating captions...")
image_files = ["sample_cat.jpg", "sample_geometric.jpg", "sample_landscape.jpg"]
test_images = [
    load_image_from_file(f"images/{f}") 
    for f in image_files 
    if os.path.exists(f"images/{f}")
]

test_captions = batch_generate_captions(test_images, processor, model)

print("\nGenerated captions:")
for i, caption in enumerate(test_captions, 1):
    print(f"{i}. {caption}")

# Analyze word frequencies
word_counts = count_words(test_captions)

print("\n" + "="*60)
print("Word Frequency Analysis")
print("="*60)
print("\nTop 5 most common words:")
for word, count in word_counts.most_common(5):
    print(f"  {word}: {count}")

print(f"\nTotal unique words: {len(word_counts)}")
print(f"Total words analyzed: {sum(word_counts.values())}")
```

### Bonus: Visualization

```python
# Bonus: Visualize word frequencies
import matplotlib.pyplot as plt

top_words = word_counts.most_common(10)
words, counts = zip(*top_words) if top_words else ([], [])

plt.figure(figsize=(10, 6))
plt.bar(words, counts, color='skyblue', edgecolor='navy')
plt.xlabel('Words', fontsize=12)
plt.ylabel('Frequency', fontsize=12)
plt.title('Top 10 Most Common Words in Captions', fontsize=14)
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

## Exercise 5 Solution: Build a Simple Caption Gallery

### Teaching Points:
- Working with file systems (os module)
- List comprehension with filtering
- Dynamic subplot creation
- File I/O for saving results

### Discussion:
- How would you handle very large directories?
- What about subdirectories?
- How could we add caching to avoid re-processing?

```python
# SOLUTION TO EXERCISE 5
# ======================

def create_caption_gallery(directory_path):
    """
    Create a gallery of images with AI-generated captions.
    
    Args:
        directory_path (str): Path to directory containing images
        
    Returns:
        list: List of (filename, image, caption) tuples
    """
    print(f"Processing images from: {directory_path}")
    
    # Step 1: Get list of all files
    if not os.path.exists(directory_path):
        print(f"Error: Directory not found: {directory_path}")
        return []
    
    all_files = os.listdir(directory_path)
    
    # Step 2: Filter for valid image files
    image_files = [f for f in all_files if is_allowed_file(f)]
    image_files.sort()  # Sort alphabetically
    
    if not image_files:
        print("No valid image files found.")
        return []
    
    print(f"Found {len(image_files)} image(s)")
    
    # Step 3: Load all images
    images = []
    for filename in image_files:
        filepath = os.path.join(directory_path, filename)
        try:
            img = load_image_from_file(filepath)
            images.append((filename, img))
        except Exception as e:
            print(f"Warning: Could not load {filename}: {e}")
    
    # Step 4: Generate captions
    print(f"\nGenerating captions for {len(images)} images...")
    just_images = [img for _, img in images]
    captions = batch_generate_captions(just_images, processor, model)
    
    # Combine results
    results = [(name, img, cap) for (name, img), cap in zip(images, captions)]
    
    # Step 5: Display in a grid
    n_images = len(results)
    if n_images == 0:
        return []
    
    # Calculate grid dimensions
    cols = min(3, n_images)
    rows = (n_images + cols - 1) // cols
    
    fig, axes = plt.subplots(rows, cols, figsize=(5*cols, 5*rows))
    
    # Handle single image case
    if n_images == 1:
        axes = [[axes]]
    elif rows == 1:
        axes = [axes]
    
    # Plot each image
    for idx, (filename, img, caption) in enumerate(results):
        row = idx // cols
        col = idx % cols
        ax = axes[row][col] if rows > 1 else axes[0][col] if cols > 1 else axes[0][0]
        
        ax.imshow(img)
        ax.axis('off')
        ax.set_title(f'{filename}\n"{caption}"', fontsize=10, wrap=True)
    
    # Hide empty subplots
    for idx in range(n_images, rows * cols):
        row = idx // cols
        col = idx % cols
        ax = axes[row][col] if rows > 1 else axes[0][col]
        ax.axis('off')
    
    plt.tight_layout()
    plt.show()
    
    # Step 6: Return results
    return results

# Test the function
print("="*60)
results = create_caption_gallery("images")
print("="*60)
print(f"\n✓ Processed {len(results)} image(s)")

# Bonus: Save to file
if results:
    output_file = "captions_output.txt"
    with open(output_file, "w") as f:
        f.write("AI Image Captioning Results\n")
        f.write("="*60 + "\n\n")
        for i, (filename, img, caption) in enumerate(results, 1):
            f.write(f"{i}. {filename}\n")
            f.write(f"   Caption: {caption}\n")
            f.write(f"   Size: {img.size[0]}x{img.size[1]}\n\n")
    
    print(f"\n✓ Results saved to: {output_file}")
```

## Additional Teaching Examples

### Example 1: Error Handling Patterns

```python
# Teaching error handling
def safe_caption_generation(image_path):
    """Demonstrate comprehensive error handling."""
    try:
        # Try to load and caption
        image = load_image_from_file(image_path)
        caption = generate_caption(image, processor, model)
        return caption
        
    except FileNotFoundError:
        return "Error: Image file not found"
    except PermissionError:
        return "Error: Permission denied to read file"
    except Exception as e:
        return f"Error: {str(e)}"

# Test with various inputs
test_paths = [
    "images/sample_cat.jpg",  # Valid
    "images/nonexistent.jpg",  # Not found
    "/root/protected.jpg"  # Permission issue (example)
]

for path in test_paths:
    result = safe_caption_generation(path)
    print(f"{path}: {result}")
```

```python
# Demonstrate batch processing efficiency
import time

# Load test images
test_images = [
    load_image_from_file(f"images/{f}") 
    for f in ["sample_cat.jpg", "sample_geometric.jpg", "sample_landscape.jpg"]
    if os.path.exists(f"images/{f}")
]

# Method 1: One by one
start = time.time()
captions_sequential = [generate_caption(img, processor, model) for img in test_images]
time_sequential = time.time() - start

# Method 2: Batch processing
start = time.time()
captions_batch = batch_generate_captions(test_images, processor, model)
time_batch = time.time() - start

print("Performance Comparison:")
print(f"  Sequential: {time_sequential:.2f} seconds")
print(f"  Batch: {time_batch:.2f} seconds")
print(f"  Speedup: {time_sequential/time_batch:.2f}x faster")
```
