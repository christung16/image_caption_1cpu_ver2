## Part 3: Loading and Processing Images

Images can come from different sources:
- Local files
- URLs from the internet
- Camera input (advanced)

Let's try loading images from files first.

````python
# Load a sample image
image_path = "images/sample_cat.jpg"
image = load_image_from_file(image_path)

# Display the image
plt.figure(figsize=(8, 6))
plt.imshow(image)
plt.axis('off')
plt.title("Sample Image")
plt.show()

print(f"Image size: {image.size[0]}x{image.size[1]} pixels")
print(f"Image mode: {image.mode}")
````

### 🖼️ Understanding Image Formats

Images can be in different modes:
- **RGB**: Red, Green, Blue (3 channels) - most common
- **RGBA**: RGB + Alpha (transparency)
- **L**: Grayscale (1 channel)

Our model works with RGB images.

---

## Part 4: Generating Captions

Now for the exciting part - let's generate a caption for our image!

````python
# Generate a caption
print("Generating caption...")
caption = generate_caption(image, processor, model)

# Display image with caption
plt.figure(figsize=(10, 6))
plt.imshow(image)
plt.axis('off')
plt.title(f'Caption: "{caption}"', fontsize=14, pad=20)
plt.show()

print(f"\n📝 Generated caption: \"{caption}\"")
````

### 🔬 How Does It Work?

The caption generation process:

1. **Image Preprocessing**: Convert image to tensors (numbers)
2. **Encoder**: Extract visual features from the image
3. **Decoder**: Generate text tokens word by word
4. **Decoding**: Convert tokens back to readable text

This all happens in milliseconds!

---

## Part 5: Comparing Different Captions

AI can generate different captions for the same image. Let's see variations!

````python
# Generate 3 different captions
print("Generating multiple caption variations...")
captions = compare_captions(image, processor, model, num_variations=3)

# Display results
print("\n📝 Caption Variations:")
for i, cap in enumerate(captions, 1):
    print(f"{i}. {cap}")

# Display image with all captions
plt.figure(figsize=(10, 8))
plt.imshow(image)
plt.axis('off')
title_text = "\n".join([f"{i}. {cap}" for i, cap in enumerate(captions, 1)])
plt.title(f"Caption Variations:\n{title_text}", fontsize=12, pad=20)
plt.show()
````

---

## Part 6: Batch Processing Multiple Images

Processing images one by one is slow. Let's process multiple images at once!

````python
# Load all sample images
image_files = ["sample_cat.jpg", "sample_geometric.jpg", "sample_landscape.jpg"]
images = []

for filename in image_files:
    img_path = os.path.join("images", filename)
    if os.path.exists(img_path):
        images.append(load_image_from_file(img_path))

print(f"Loaded {len(images)} images")

# Generate captions for all images at once
captions = batch_generate_captions(images, processor, model)

# Display results
fig, axes = plt.subplots(1, len(images), figsize=(15, 5))
if len(images) == 1:
    axes = [axes]

for i, (img, cap) in enumerate(zip(images, captions)):
    axes[i].imshow(img)
    axes[i].axis('off')
    axes[i].set_title(f'"{cap}"', fontsize=10)

plt.tight_layout()
plt.show()
````
